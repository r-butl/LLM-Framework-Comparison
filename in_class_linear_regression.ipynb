```python
import numpy as np
import plotnine as pn
import pandas as pd
import patsy as pt
from sklearn.linear_model import LinearRegression
```

```python
carnivora_url = "https://raw.githubusercontent.com/roualdes/data/refs/heads/master/carnivora.csv"
df = pd.read_csv(carnivora_url)
```

```python
df.columns
```

```python
plot = pn.ggplot(data=df, mapping=pn.aes(x='SW', y='SB')) + \
    pn.geom_point() + \
    pn.scale_x_log10() + \
    pn.scale_y_log10() + \
    pn.geom_smooth(method='lm', se=False)

print(plot)
```

```python
y, X = pt.dmatrices("SB ~ SW", df)
model = LinearRegression(fit_intercept=False).fit(X, y)
```

```python
B0, B1 = model.coef_[0]
```

```python
B0, B1
```

```python
y = np.log10(df['SB'])
x = np.log10(df['SW'])
y_hat = x * B1 + B0
```

```python
x
```

```python
y
```

```python
y_hat
```

```python
residual = y_hat - y 
```

```python
for r in residual:
    print(r)
```
